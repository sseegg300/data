```python
#Series객체 생성(1)    #Ctrl + . :funtion list help
temp = pd.Series([1.2,3.5,5.8,13.3,15.7])
print(temp)
#type(temp)
```

### DF 생성(1) <- 리스트 구조


```python
#DataFrame 생성(1)
data = [[15, 160.5, '서울'],
        [16, 158.3, '대전'],
        [20, 175.0, '대구'],
        [19, 178.4, '부산'],
        [23, 185.0, '제주']]

idx = ['홍길동', '이순신', '유관순', '박지훈', '박정민']

cols = ['나이', '키', '사는곳']

df = pd.DataFrame(data,
             index = idx,
             columns=cols)
df
```

```python
type(df)
```

```python
df['나이']  #df[컬럼명] - Series타입이라 df[]안에 컬럼명을 str로 바로 입력가능
# type(df['나이'])
```

```python
df[['나이','키']]  #df[[컬럼명2,컬럼명2]] - DataFrame타입이라 df[]안에 컬럼명을 리스트형식으로 입력
#type(df[['나이','키']])
```

```python
#행(열)별 자료 검색 방법 - loc()
#df.loc[행인덱스,열인덱스]
df.loc['홍길동',:]  #df.loc['홍길동']
#type(df.loc['홍길동'])
#df.loc['홍길동',['나이','키']]
```

## 8️⃣그룹화

* 동일한 값으로 그룹을 만들어 합계,평균등 통계값을 계산하기 위함.


```python
import pandas as pd
# df = pd.read_csv('https://raw.githubusercontent.com/sseegg300/data/refs/heads/main/k-scores.csv')
# df['번호']=['1번','2번','3번','4번','5번','6번','7번']
# df.set_index('번호', inplace=True)

url = 'https://raw.githubusercontent.com/sseegg300/data/refs/heads/main/k-scores.csv'
df = pd.read_csv(url).assign(번호=['1번','2번','3번','4번','5번','6번','7번']).set_index('번호')
df
```

```python
# 학년별로 그룹화
df.groupby('학년')
```

```python
#1학년 학생만 가져오기
df.groupby('학년').get_group(1)
```

```python
#학년별로 계산 가능한 컬럼의 평균 값을 확인
#Pandas의 mean()은 기본적으로 숫자 데이터만 계산하려고 하지만,
#명시적으로 설정을 해주지 않으면 문자열을 어떻게 처리할지 몰라 에러
df.groupby('학년').mean(numeric_only=True)
```

```python
#그룹별 크기확인
df.groupby('학년').size()
df.groupby('학년').size()[3]    #3학년의 크기
```

```python
#학년별 국어점수 평균
df.groupby('학년')['국어'].mean()
df.groupby('학년')['국어'].mean().astype(int)
```

```python
#학년별 국영수 평균확인
df.groupby('학년')[['국어','수학','영어']].mean().astype(int)  #평균을 정수로
df.groupby('학년')[['국어','수학','영어']].mean().round(1)     #평균을 소수첫째자리까지
```

```python
#학교별 학년별 평균 구하기/국어점수순
df.groupby(['학교','학년']).mean(numeric_only=True)
df.groupby(['학교','학년']).mean(numeric_only=True).sort_values('국어')
```

```python
#학교별 국어평균
df.groupby('학교')['국어'].mean()
df.groupby('학교')['국어'].mean().sort_values(ascending=False)  #역순 정렬
```

```python
#학교별 동아리 데이터의 갯수 확인(예:부산고는 3개의 동아리가 있음 )
df.groupby('학교')['동아리'].count()
```
