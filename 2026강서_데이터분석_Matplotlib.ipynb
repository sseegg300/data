## 2️⃣꺽은선 그래프

```python
# 꺾은선 그래프 - 스타일
#(1) 라인 스타일,색상,굵기,투명도
#(2) marker 모양,크기,색
import matplotlib.pyplot as plt
x = ['1월', '2월', '3월', '4월', '5월']
y = [56, 55, 54, 55, 57]  #단위 : kg

plt.plot(x, y,
         color='plum',
         linestyle='--',
         linewidth=2,
         alpha = 0.3,
         marker='s',
         markersize=10,
         markeredgecolor='red',
         markerfacecolor='gold'
)

plt.title('몸무게 변화')
plt.ylabel("kg")
plt.grid(True)
```

```python
#포맷으로 간단하게 스타일 정하는 방법
plt.plot(x,y,'ro--')   #색상-마커모양- 선스타일
```

```python
#축약어로 그리는 방법(구글검색: matplotlib markerfacecolor mfc)
plt.plot(x,y,'bo--', mfc='pink', mec='green',ms=15)
plt.grid(alpha=0.5,linestyle=':')
pass
```

```python
#maker에 수치를 직접 표시하는 방법(CANVAS 설정 없이 하는 방법) -  for 문 사용

plt.plot(x, y,
         color='plum',
         marker='s',
         linestyle='--',
         linewidth=2)

for i in range(len(x)):
  plt.text(x[i], y[i] + 0.2,      # 좌표 (x, y + 약간의 여백)
             y[i],             # 표시할 텍스트
             fontsize=12,
             color='black')           # 세로 정렬 (bottom, top, center)

# enumerate(y): y 리스트에서 **몇 번째인지(idx)**와 **그 값(txt)**을 동시에 꺼내옵
# for i , txt in enumerate(y):
#   plt.text(x[i],y[i]+0.2,txt,fontsize=12, color='red')


plt.title('몸무게 변화')
plt.xlabel('일자')
plt.ylabel('kg')
plt.grid(True)
```

## 3️⃣막대그래프

`plt.bar(x데이터,y데이터)`
`plt.barh(x데이터,y데이터)`

```python
# x, y축에 들어갈 데이터
x = ['1월', '2월', '3월', '4월', '5월']
y = [80, 78, 73, 75, 76]

plt.bar(x,y,color='pink')
plt.title('월별 몸무게의 변화')
plt.grid(True)
```

```python
# 막대별 다양한 컬러(컬러 리스트 활용)
x = ['1월', '2월', '3월', '4월', '5월']
y = [80, 78, 73, 75, 76]
c = ['r','g','b','pink','plum']

plt.bar(x,y, color = c, alpha = 0.5)
plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
```

```python
#y값 제약으로 비교하기 쉽게하는 방법 .ylim()
plt.bar(x,y, color = c, alpha = 0.5)
plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
plt.ylim(60,85)
```

```python
#(그래프 기본) 축 value의 회전 표시 : .xticks() .yticks()
plt.bar(x,y, color = c, alpha = 0.5, width=0.5)
plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
plt.ylim(60,85)
plt.xticks(rotation=45)
plt.yticks(rotation=45)
pass
```

```python
# 막대그래프 위에 글자 넣기(1)
plt.bar(x,y, color = c, alpha = 0.5, width=0.5)
plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
plt.ylim(60,85)
plt.xticks(rotation=45)
plt.yticks(rotation=45)

for i, txt in enumerate(y):
  plt.text(x[i],y[i]+0.2,txt, ha='center', color='green', size=8)

pass
```

```python
#🎈막대그래프 위에 글자 넣기(2) - bar 각각의 데이터 활용 방법
bar = plt.bar(x,y, color = c, alpha = 0.5, width=0.5)   #bar[0],bar[1],bar[2]...

for i,rect in enumerate(bar):
  plt.text(i, rect.get_height()+0.3, y[i], ha = 'center', color='purple')

plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
plt.ylim(60,85)
pass
```

```python
#🎈막대그래프 패턴 넣기
bar = plt.bar(x,y, color = c, alpha = 0.5, width=0.5)   #bar[0],bar[1],bar[2]...

for i,rect in enumerate(bar):
  plt.text(i, rect.get_height()+0.3, y[i], ha = 'center', color='purple')

plt.title('월별 몸무게의 변화')
plt.grid(linestyle=':')
plt.ylim(60,85)
pass
```

```python
# 샘플데이터  : 100명 중 혈액형별 인원의 비율
# 파이차트
#plt.pie?
blood_type=['AA','AO','BB','BO','O','AB']
count=[20,25,20,15,15,5]

for i in range(len(blood_type)):
  blood_type[i] = blood_type[i]  + "형"

#plt.pie(count)
#plt.pie(count,labels=blood_type)
#plt.pie(count,labels=blood_type, autopct='%.1f')
#plt.pie(count,labels=blood_type, autopct='%.1f%%') # %붙히기
#plt.pie(count,labels=blood_type, autopct='%.1f%%', explode=[0.05,0,0,0])
#plt.pie(count,labels=blood_type, autopct='%.1f%%', explode=[0.05]*4)
#plt.pie(count,labels=blood_type, autopct='%.1f%%', explode=[0.05,0,0,0],startangle=90,counterclock=False)
#plt.legend()   #loc=(1,1) : 우측 맨 위 위치
#plt.legend(loc=(1,0.7))

plt.pie(count,
        labels=blood_type,
        autopct='%1.1f%%',
        explode=[0.05,0,0,0,0,0],
        startangle=90,
        counterclock=False
        )
plt.legend(loc=(1,0.7))
plt.title('혈액형별 인원의 비율')
plt.show()
```

```python
blood_type=['AA','AO','BB','BO','O','AB']
count=[20,25,20,15,15,5]
colors = ['#e89bd0','#70cf89','#f2f7ab','#99b3e8','#a9ed74','#b57d84']
for i in range(len(blood_type)):
  blood_type[i] = blood_type[i]  + "형"

plt.pie(count,
        labels=blood_type,
        colors=colors,
        autopct='%1.1f%%',
#  explode=[0.05,0,0,0,0,0],
        startangle=90,
        counterclock=False,
        wedgeprops={'width': 0.5, 'edgecolor': 'w', 'linewidth': 5}, #전체 파이두께(1:not 도넛)
        textprops={'fontsize': 10, 'color': '#0a55f0', 'fontweight': 'bold'},
        pctdistance=0.7
        )
plt.legend(loc=(1.1,0.6))
plt.title('혈액형별 인원의 비율')
plt.show()
```

```python
# 🎈함수 적용  낮은% 레이블은 나타내지 않기
def skip_autopct(p):
  return f'{p:.1f}%' if p >=10 else ''

 #return ('%.1%%' %p) if p >=10 else ''
 #return'{.1f}%'.format(p) if p > 10 else ''

plt.pie(count,
        labels=blood_type,
        colors=colors,
        startangle=90,
        counterclock=False,
        autopct=skip_autopct,
        wedgeprops={'width': 0.5, 'edgecolor': 'w', 'linewidth': 5}, #전체 파이두께(1:not 도넛)
        textprops={'fontsize': 12, 'color': '#0a55f0', 'fontweight': 'bold'},
        pctdistance=0.7
        )
plt.legend(loc=(1.1,0.6))
plt.title('혈액형별 인원의 비율')
plt.show()
```

### seaborn 항공 승객 데이터 활용

```python
#seaborn 항공 승객 데이터
import seaborn as sns

flights = sns.load_dataset('flights')
flights.info()
flights.columns

flights.columns = ['년도', '월','승객수']
flights.columns
```

```python
# 특정년도의 월별 승객수 선 그래프
f_1960 = flights[flights['년도'] == 1960]
f_1956 = flights[flights['년도'] == 1956]
f_1950 = flights[flights['년도'] == 1950]
plt.plot(f_1960['월'],f_1960['승객수'],
         marker='o',
         color='red',
         label='1960년'
)
plt.plot(f_1956['월'],f_1956['승객수'],
         marker='o',
         color='green',
         label='1956년'
)
plt.plot(f_1950['월'],f_1950['승객수'],
         marker='o',
         color='blue',
         label='1950년'
)
plt.title('<월별 승객수>',pad=20,fontsize=15)
plt.xlabel('월')
plt.ylabel('승객수')
plt.grid(True, linestyle='--', alpha=0.5) #alpha는 투명도 0.0에서 1.0
plt.legend(title='연도별', loc='upper left', fontsize=10, shadow=True)
plt.show() #:화면이 깔끔하게 정리됨.
```

```python
# 특정년도의 월별 승객수 선 그래프(수치표현)
f_1960 = flights[flights['년도'] == 1960]
f_1956 = flights[flights['년도'] == 1956]
f_1950 = flights[flights['년도'] == 1950]
plt.plot(f_1960['월'],f_1960['승객수'],
         marker='o',
         color='red',
         label='1960년'
)

for i,txt in enumerate(f_1956['승객수']):
  plt.text(f_1956['월'].iloc[i], txt + 15, txt, fontsize=7)

plt.plot(f_1956['월'],f_1956['승객수'],
         marker='o',
         color='green',
         label='1956년'
)

for i,txt in enumerate(f_1960['승객수']):
  plt.text(f_1960['월'].iloc[i], txt + 15, txt, fontsize=7)

plt.plot(f_1950['월'],f_1950['승객수'],
         marker='o',
         color='blue',
         label='1950년'
)

for i,txt in enumerate(f_1950['승객수']):
  plt.text(f_1950['월'].iloc[i], txt + 15, txt, fontsize=7)

plt.title('<월별 승객수>',pad=20,fontsize=15)
plt.xlabel('월')
plt.ylabel('승객수')
plt.grid(True, linestyle='--', alpha=0.5) #alpha는 투명도 0.0에서 1.0
plt.legend(title='연도별', loc='upper left', fontsize=10, shadow=True)
plt.show() #:화면이 깔끔하게 정리됨.
```

```python
# 학생별 평균점수 컬럼 추가
avg = round((df['국어']+df['영어']+df['수학'])/3,2)
type(avg)
df['평균']= avg
df
```

```python
ott_data = {
    '콘텐츠명': ['오징어 게임 2', '더 글로리', '무빙', '피지컬: 100', '환승연애', '흑백요리사', '카지노'],
    '플랫폼': ['Netflix', 'Netflix', 'Disney+', 'Netflix', 'TVING', 'Netflix', 'Disney+'],
    '개봉년도': [2024, 2022, 2023, 2023, 2021, 2024, 2022],
    '시청시간': [16.5, 6.2, 5.8, 2.1, 1.5, 4.8, 3.2],    #(단위:억)
    '평점': [9.2, 9.5, 9.7, 8.8, 9.1, 9.4, 8.9]
}

df = pd.DataFrame(ott_data)
df
```

```python
#✅[문제1] 막대그래프 그리기(X:콘텐츠명, Y:평점)
plt.figure(figsize=(5,3))
plt.bar(df['콘텐츠명'],df['평점'],color='pink')

pass
```

```python
#✅[문제2] 그래프 제목,x,y축 라벨을 적용해서 막대그래프를 완성 하세요.

# 제목:<콘텐츠별 평점>
# x축 label: 콘텐츠명(30도 회전)
# y축 label: 평점

plt.figure(figsize=(5,3))
plt.bar(df['콘텐츠명'],df['평점'],color='pink')
plt.xlabel('콘텐츠명')
plt.ylabel('평점')
plt.xticks(rotation=30, size=7)
plt.title('<콘텐츠별 평점>')

for i,value in enumerate(df['평점']):
  plt.text(df['콘텐츠명'].iloc[i],value + 0.3, value,ha='center', size=8)
```

```python
#✅[문제3] 개봉년도별 평균시청시간 변화를 선그래프로 그려보세요.
#[문제3-1] 개봉년도별로 groupby를 해서 평균값(.mean())을 먼저 구하시오
df_group = df.groupby('개봉년도').mean(numeric_only=True)
df_group
```

```python
#✅[문제3] 개봉년도별 평균시청시간 변화를 선그래프로 그려보세요.
#[문제3-2] 선그래프 그리기
plt.plot(df_group.index,df_group['시청시간'])
pass
```

```python
#✅[문제3] 개봉년도별 평균시청시간 변화를 선그래프로 그려보세요.
#[문제3-3]  위 선그래프 아래 조건에 맞게 그려주세요.

# 제목:<년도별 평균시청시간>
# x축 label: 년도
# y축 label: 시간(단위:억)
# marker:'o'
# x축눈금(2021,2022,2023,2024)

plt.plot(df_group.index,df_group['시청시간'],marker='o',markersize=10)
plt.title('<년도별 평균시청시간>')
plt.xlabel('년도')
plt.ylabel('시간(단위:억)')
plt.xticks([2021,2022,2023,2024])
pass
```
